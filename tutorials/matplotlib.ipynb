# Introduction to Matplotlib
---

- Library to display and visualize data
- Serves also as a base library for many other libraries such as Seaborn, Plotly, pandas 

https://matplotlib.org/stable/tutorials/index

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# First we need some data
N   = 1000
arr = np.cumsum(np.random.normal(0,1,N))
```

**Simple Line Plots**
- In general a plot consists of a figure object or canvas / background
- An Axes object serving as the coordinate system on which we display the data
- An the data itself

```python
# Make a figure object called fig
fig = plt.figure()

# Add a coordinate system called ax
ax = fig.add_subplot()

# Display the data, and label it for legend
ax.plot(arr, label='frist measurement')

# Define x, y Labels
ax.set_xlabel("Time")
ax.set_ylabel("Data")

# Add a legend
ax.legend()

# Add grids
ax.grid()
```

**Multi-Plots**
- We might want to display different runs in a single plot
- Or in multiple plots using `fig.add_subplot(n_row,n_cols,nth_plot)`

```python
# Again some data
N   = 1000
arr = np.cumsum(np.random.normal(0,1,size=(N,4)),axis=0)
```

```python
# Make a figure object called fig
fig = plt.figure()

# Add a coordinate system called ax
ax = fig.add_subplot()

# Display the data, and label it for legend
for i in range(4):
    ax.plot(arr[:,i], label=f'Run {i}')

# Define x, y Labels
ax.set_xlabel("Time")
ax.set_ylabel("Data")

# Add a legend
ax.legend()

# Add grids
ax.grid()
```

```python
# Make a figure object called fig with a constrained layout (best for multiplots)
fig = plt.figure(layout='constrained')

# Iterate over each run
for i in range(4):
    # And make a new subplot in a 2x2 grid of plots with i+1 as the plot count
    ax = fig.add_subplot(2,2,i+1)
    ax.plot(arr[:,i], label=f'Run {i}')

    # Define x, y Labels
    ax.set_xlabel("Time")
    ax.set_ylabel("Data")

    # Add a legend
    ax.legend()

    # Add grids
    ax.grid()
```

- Use Style Sheets for different plot styles

https://matplotlib.org/stable/gallery/style_sheets/style_sheets_reference.html

```python
with plt.style.context("bmh"):

    # Make a figure object called fig
    fig = plt.figure()

    # Add a coordinate system called ax
    ax = fig.add_subplot()

    # Display the data, and label it for legend
    for i in range(4):
        ax.plot(arr[:,i], label=f'Run {i}')

    # Define x, y Labels
    ax.set_xlabel("Time")
    ax.set_ylabel("Data")

    # Add a legend
    ax.legend()

    # Add grids
    ax.grid()
```

**Different Plot Functions**

```python
# Data
N       = 10000
arr1    = np.random.normal(0,5,size=(N))
arr2    = np.random.normal(0,1,size=(N))
arr3    = np.random.normal(4,1,size=(N))
```

```python
with plt.style.context("fivethirtyeight"):

    fig = plt.figure()
    ax  = fig.add_subplot()
    h   = ax.hist(arr1, bins=100, label='Run 1')
    h   = ax.hist(arr2, bins=100, label='Run 2')
    h   = ax.hist(arr3, bins=100, label='Run 3')
    ax.legend()
```

```python
# Create sample 2D data for phase plot
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y)

# Create the phase plot
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot()
im = ax.imshow(Z, extent=[-5, 5, -5, 5], origin='lower', cmap='viridis')

# Add colorbar and labels
plt.colorbar(im)
ax.set_xlabel('X')
ax.set_ylabel('Y')
```
