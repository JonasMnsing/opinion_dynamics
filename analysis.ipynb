```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

Load results which has a shape of (50,1000,80) as there are 50 runs with 1000 time steps for 80 agents

```python
arr = np.load("opinions.npy")
arr.shape
```

We could e.g. select time steps [0,100,999]

```python
arr_0   = arr[:,0,:].copy()
arr_100 = arr[:,100,:].copy()
arr_999 = arr[:,999,:].copy()
```

For a distribtion at the last time step we can just collapse our 2D array into a 1D array by using `flatten`

```python
fig,ax  = plt.subplots(dpi=100)
ax      = sns.kdeplot(arr_0.flatten(), ax=ax)
ax      = sns.kdeplot(arr_100.flatten(), ax=ax)
ax      = sns.kdeplot(arr_999.flatten(), ax=ax)
ax.set_xlabel('Opinions')
ax.set_xlim(-1,1)
```

### At the end we get polarization. There are three major opinions [-1,0,1]
