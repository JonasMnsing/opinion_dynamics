```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.cm import ScalarMappable 
from matplotlib.colors import Normalize

plt.style.use("seaborn-v0_8-whitegrid")
```

```python
arr     = np.load("../../data/social_impact/er_noise.npy", allow_pickle=True)
params  = np.load("../../data/social_impact/er_noise_params.npy", allow_pickle=True)
noise_l = params[:, 3]
avg_op  = np.mean(np.abs(arr), axis=1)
avg_op.shape
```

### Different Trajectories

```python
fig, ax = plt.subplots(dpi=100)
for i in range(50):
    ax.plot(np.abs(arr[0,i,:]))
ax.set_xlabel("Time step")
ax.set_ylabel("Average opinion")
```

### Ensemble Mean

```python
fig, ax = plt.subplots(dpi=100)
cm      = plt.get_cmap('cividis')
norm    = Normalize(vmin=min(noise_l), vmax=max(noise_l))
sm      = ScalarMappable(cmap=cm, norm=norm)
sm.set_array([])

for i, noise in enumerate(noise_l):
    ax.plot(avg_op[i,:], color=cm(norm(noise)))

ax.set_xlabel("Time step")
ax.set_ylabel("Average opinion")

cbar = fig.colorbar(sm, ax=ax, label='Noise')
```
