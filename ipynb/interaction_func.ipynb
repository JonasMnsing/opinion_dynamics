```python
import numpy as np
import matplotlib.pyplot as plt
import scienceplots
plt.style.use(["science","bright","grid"])
```

```python
def social_impact(xi, xj, w=1, p=1, s=1):
    return w*(p*(1-xi*xj) - s*(1+xi*xj))
def social_impact3(xi, xj, w=1, s=1):
    return w*(s*(1+xi*xj) - (1-s)*(1-xi*xj))
def social_impact2(xi, xj, w=1, p=1):
    return w*p*xi*xj
def social_impact3(xi, xj, w=1, s=1):
    return w*(s*(1+xi*xj) - (1-s)*(1-xi*xj))
def distance_based(xi, xj, w=1, p=1):
    return w*p*(xj-xi)


xi_vals = [-1.0,-0.5,-0.2,0.0,0.2,0.5,1.0]
xj_vals = np.linspace(-1,1,1001,endpoint=True)
```

```python
fig = plt.figure(dpi=200, layout='constrained')
fig.set_figwidth(fig.get_figwidth()*2)
ax1 = fig.add_subplot(1,2,1)
ax2 = fig.add_subplot(1,2,2)

for xi in xi_vals:
    ax1.plot(xj_vals, social_impact(xi, xj_vals), label=f'$x_i = {xi}$')
    ax2.plot(xj_vals, -np.tanh(2*xi*social_impact(xi, xj_vals)), label=f'$x_i = {xi}$')

ax1.set_xlabel('$x_j$')
ax1.set_ylabel('$I_i \sim [1+x_ix_j]-[1-x_ix_j]$')
ax1.set_title("$p=1$, $s=1$")
ax2.set_xlabel('$x_j$')
ax2.set_ylabel('$x_i = \\tanh(\\beta \cdot x_i \cdot I_i)$')
ax2.legend(ncol=1, bbox_to_anchor=(1.55,1.04), loc='upper right')
fig.savefig('plots/funcs/2_parameter_social_impact_p1_s1.png')
```
