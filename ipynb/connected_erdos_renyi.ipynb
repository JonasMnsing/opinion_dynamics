```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
params  = np.load("../data/voter/params.npy")
arr     = np.load("../data/voter/m_trajs.npy")
p_cons  = [p[0] for p in params]
p_noise = [p[1] for p in params]
means   = [np.mean(np.abs(arr[i,:,-1])) for i in range(len(params))]
df      = pd.DataFrame([p_cons,p_noise,means],index=['$p_c$','$p_n$','$y$']).T
df.head()
```

```python
with plt.style.context("ggplot"):
    fig = plt.figure(dpi=200)
    ax  = fig.add_subplot()
    ax  = sns.lineplot(data=df, x='$p_n$', y='$y$', hue='$p_c$', palette='cividis', ax=ax)
    
```
